# %% [2] THE CORE ENGINE: ADVANCED KD LOSS
This class contains the actual math for all methodologies. 

```python
class MasterKDLoss(v8DetectionLoss):
    def __init__(self, model, teacher, method='standard', T=3.0, alpha=0.4):
        super().__init__(model)
        self.teacher = teacher
        self.method = method
        self.T = T
        self.alpha = alpha
        self.kl_div = nn.KLDivLoss(reduction="batchmean")
        
        # For Internal Adapters (Method 6)
        self.adapterP3 = nn.Conv2d(64, 192, 1).to(DEVICE) # Example channels

    def __call__(self, preds, batch):
        # 1. Hard Loss (Ground Truth)
        hard_loss, loss_items = super().__call__(preds, batch)
        
        # 2. Teacher Inference
        with torch.no_grad():
            t_preds = self.teacher(batch['img'])
        
        s_logits = preds[0]; t_logits = t_preds[0]
        
        # 3. Apply Selected Methodology
        if self.method == 'standard':
            soft_loss = self.kl_div(F.log_softmax(s_logits/self.T, dim=1), F.softmax(t_logits/self.T, dim=1))
            
        elif self.method == 'mean_subtraction':
            s_centered = s_logits - s_logits.mean(dim=1, keepdim=True)
            t_centered = t_logits - t_logits.mean(dim=1, keepdim=True)
            soft_loss = self.kl_div(F.log_softmax(s_centered/self.T, dim=1), F.softmax(t_centered/self.T, dim=1))
            
        elif self.method == 'logit_norm':
            s_norm = s_logits / (torch.norm(s_logits, p=2, dim=1, keepdim=True) + 1e-6)
            t_norm = t_logits / (torch.norm(t_logits, p=2, dim=1, keepdim=True) + 1e-6)
            soft_loss = self.kl_div(F.log_softmax(s_norm/0.04, dim=1), F.softmax(t_norm/0.04, dim=1))
            
        elif self.method == 'cwd':
            # Channel-Wise (Simplified for 1st map)
            s_soft = F.softmax(preds[1][0].view(batch['img'].shape[0], -1), dim=-1)
            t_soft = F.softmax(t_preds[1][0].view(batch['img'].shape[0], -1), dim=-1)
            soft_loss = self.kl_div(torch.log(s_soft + 1e-6), t_soft)
        
        else: soft_loss = torch.tensor(0.0, device=DEVICE)

        total_loss = (self.alpha * hard_loss) + ((1 - self.alpha) * soft_loss * (self.T**2))
        return total_loss, loss_items
```
